## Esercizio 1: Somma di due numeri
Scrivi una funzione che prende in input due numeri e restituisce la loro somma.

```python
def somma(a, b):
    # Scrivi la somma di due addendi
  
    return a + b

# Test della funzione
print(somma(3, 5))  # Dovrebbe stampare 8
```

## Esercizio 2: Pari o Dispari
Scrivi una funzione che verifica se un numero intero è pari o dispari. Deve restituire `True` se è pari, `False` altrimenti.

```python
def e_pari(numero):
    # Trascrizione di un numero: True se pari, False se è dispari\n",
  
  if numero % 2 == 0 :
    return True
  else :
    return False
    
# Test della funzione
print(e_pari(4))  # True
print(e_pari(7))  # False
```

## Esercizio 3: Calcolo dell'area di un cerchio
Scrivi una funzione che calcola l'area di un cerchio dato il suo raggio. Usa `pi` dal modulo `math`.

```python
import math

def area_cerchio(raggio):
    # Calcolo dell'area di un cerchio
  
    return math.pi * raggio **2

# Test della funzione
print(area_cerchio(5))  # Circa 78.54
```

## Esercizio 4: Massimo di tre numeri
Scrivi una funzione che prende tre numeri e restituisce il maggiore tra essi (senza usare la funzione integrata `max()`).

```python
def massimo_di_tre(a, b, c):
    # Scrivi qual è il numero maggiore tra i tre inseriti in input

  if a > b and a > c:
    return a
  else:
    if a < c and b < c:
      return c
    else:
      return b

# Test della funzione
print(massimo_di_tre(10, 25, 15))  # 25
```

## Esercizio 5: Inversione di una stringa
Scrivi una funzione che prende una stringa in input e restituisce la stringa invertita.

```python
def inverti_stringa(testo):
    # Riscrivi il testo in input emettendolo al contrario
  
    reverse = ""
    for i in range (len(testo), 0, -1):
      print(f"{i-1}: {testo[i -1]}")
      reverse += testo[i -1]
    return reverse
  
# Test della funzione
print(inverti_stringa("python"))  # "nohtyp"
```

## Esercizio 6: Conta le vocali
Scrivi una funzione che conta il numero di vocali (a, e, i, o, u) presenti in una stringa data (ignorando maiuscole/minuscole).

```python
def conta_vocali(testo):
    # Conta quante vocali ci sono nel testo dato in input

  count = 0
  for c in testo:
    if c in ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']:
      count += 1
  return count

# Test della funzione
print(conta_vocali("Programmare in Python è bello"))
```

## Esercizio 7: Tabellina di un numero
Scrivi una funzione che stampa la tabellina di un numero da 1 a 10.

```python
def stampa_tabellina(n):
    # Scrivi una funzione che scriva la tabellina di un numero da 1 a 10

  elenco = []
  for i in range(1, 11):
    p = n * i
    elenco += [p]
  return elenco

# Test della funzione
print(stampa_tabellina(4))
```

## Esercizio 8: Somma degli elementi di una lista
Scrivi una funzione che prende una lista di numeri e restituisce la somma di tutti gli elementi (senza usare la funzione integrata `sum()`).

```python
def somma_lista(lista):
    # Somma degli elementi di una lista
    
  tot = 0
  for lista in lista :
    tot += lista
  return tot

# Test della funzione
print(somma_lista([1, 2, 3, 4, 5]))  # 15
```

## Esercizio 9: Filtrare i numeri pari da una lista
Scrivi una funzione che prende una lista di numeri interi e restituisce una nuova lista contenente solo i numeri pari.

```python
def filtra_pari(lista):
    # Trascrivi solo i numeri pari di una lista
  
  pari = []
  for numero in lista:
    if numero % 2 == 0:
      pari += numero
  return pari
  
# Test della funzione
print(filtra_pari([1, 2, 3, 4, 5, 6]))  # [2, 4, 6]
```

## Esercizio 10: Palindromo
Scrivi una funzione che verifica se una stringa è palindroma (si legge uguale da sinistra a destra e viceversa, ignorando eventuali spazi e differenze tra maiuscole e minuscole).

```python
def e_palindromo(testo):
    # Verifica se una stringa è palindroma oppure no
  reverse = ""
  for i in range (len(testo), 0, -1):
    print(f"{i-1}: {testo[i -1]}")
    reverse += testo[i -1]
  if reverse == testo:
      testo = True
  else:
      testo = False
  return testo

# Test della funzione
print(e_palindromo("anna"))       # True
print(e_palindromo("radar"))      # True
print(e_palindromo("python"))     # False
```
