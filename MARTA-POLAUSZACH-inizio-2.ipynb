### Esercizio 1: Conversione di temperatura (Celsius to Fahrenheit)
Scrivi una funzione che converte una temperatura da gradi Celsius a Fahrenheit usando la formula:
F = C * 9/5 + 32

```python
def celsius_to_fahrenheit(celsius):
    # Funzione che converte i gradi Celsius in Fahrenheit
  
  d = {}
  f = celsius * 9 / 5 + 32
  d[celsius] = f
  print(f"{celsius} * 9 / 5 + 32")
  return d

# Test della funzione
print(celsius_to_fahrenheit(0))    # Atteso: 32.0
print(celsius_to_fahrenheit(100))  # Atteso: 212.0
```

### Esercizio 2: Controllo del voto scolastico
Scrivi una funzione che prende come parametro un voto numerico (da 1 a 10) e restituisce:
- `"Insufficiente"` se il voto è minore di 6
- `"Sufficiente"` se è compreso tra 6 e 7 (inclusi)
- `"Ottimo"` se è strettamente maggiore di 7

```python
def giudizio_voto(voto):
    # Scrivi se il voto assegnato è insufficiente, sufficiente o ottimo

  if voto < 6 and voto >= 1:
    voto = "Insufficiente"
  else:
    if voto >= 6 and voto <= 7:
      voto = "Sufficiente"
    else:
        if voto > 7 and voto <= 10:
          voto = "Ottimo"
  print(f"Il voto è {voto}")
  return voto
    
# Test della funzione
print(giudizio_voto(5.5))  # Atteso: Insufficiente
print(giudizio_voto(6.5))  # Atteso: Sufficiente
print(giudizio_voto(9))    # Atteso: Ottimo
```

### Esercizio 3: Calcolo del perimetro e area di un rettangolo
Scrivi una funzione che accetta base e altezza di un rettangolo e restituisce una tupla con perimetro e area: `(perimetro, area)`.

```python
def geometria_rettangolo(base, altezza):
    # Calcola il perimetro e l'area di un rettangolo

    perimetro = 2 * base + 2 * altezza
    area = base * altezza
    return (perimetro, area)

# Test della funzione
perimetro, area = geometria_rettangolo(4, 5)
print(f"Perimetro: {perimetro}, Area: {area}")  # Atteso: Perimetro: 18, Area: 20
```
